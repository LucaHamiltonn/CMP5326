# Python Regex example

```python
import re

text = "The date is 10/10/2023; 07788999871; \\n hello world!"
pattern = r"(?P<twodigit>\b\d\d\b)" #r is used for raw string
print("The pattern is: "+ pattern)
print("The text is: "+text)

match=re.search(pattern,text)
#print (match)

if match:
  print("find two digit in text:", match.group ("twodigit"))
```

```python
import re

txt = "The rain in Spain"
x = re.sub(r"\s", r"*", txt , 1)
print(x)
```

# **The Search() Function**


The object x returned by re.search() is a **Match object**, which contains information about the search and the result. Here are some commonly used properties and methods that a Match object has:

Properties:

    string: The string passed into the function.
    re: The regular expression object.
    pos: The start position of the search in the string.
    endpos: The end position of the search in the string.
    lastindex: The integer index of the last captured capturing group.

Methods:

    group(): Returns one or more subgroups of the match.
    groups(): Returns a tuple containing all the subgroups of the match.
    groupdict(): Returns a dictionary containing all the named subgroups of the match.
    start(): Returns the starting position of the match.
    end(): Returns the ending position of the match.
    span(): Returns a tuple containing the (start, end) positions of the match.

```python
import re

txt = "The rain in Spain"
x = re.search(r"\s", txt)

print("The first white-space character is located in position:", x.start())
print("the string: ",x.string)
print("the pattern:", x.re)
print("the start pos:", x.start())
print("the end pos:", x.end())

```

```python
import re

txt = "The rain in Spain"
x = re.search("rain", txt)

# Using some properties and methods
print("String:", x.string)
print("Start position:", x.start())
print("End position:", x.end())
print("Span:", x.span())
```

```python
import re

# Capturing group example
pattern1 = r'(\d{2}\s|\d{3}\s)'
matches1 = re.findall(pattern1, "123 45 67")
print("Capturing group:", matches1)  # Output: ['123', '45', '67']

# Non-capturing group example
pattern2 = r'(?:\d{2}\s|\d{3}\s)'
matches2 = re.findall(pattern2, "123 45 67")
print("Non-capturing group:", matches2)  # Output: ['123', '45', '67']
```
